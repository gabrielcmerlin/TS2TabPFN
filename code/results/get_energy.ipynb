```python
import csv
import pandas as pd
from pathlib import Path
```

# TSC

```python
base_dir = Path("../../tsc/outputs/codecarbon_logs")

POWER_FACTOR = 3_600_000  # kWh -> Joules

results = {}  # (dataset, method) -> power

if not base_dir.exists():
    print(f"Directory not found: {base_dir}")
else:
    for dataset_dir in sorted(base_dir.iterdir()):
        if not dataset_dir.is_dir():
            continue

        dataset_name = dataset_dir.name
        print(f"\nDataset: {dataset_name}")

        results[dataset_name] = {}

        for run_dir in sorted(dataset_dir.iterdir()):
            if not run_dir.is_dir():
                continue

            method_name = run_dir.name
            print(f"\n  Method: {method_name}")

            csv_files = list(run_dir.glob("*.csv"))

            if not csv_files:
                print("    No CSV file found")
                continue

            method_energy = 0.0
            method_time = 0.0

            for csv_file in csv_files:
                print(f"\n    CSV: {csv_file.name}")

                with open(csv_file, newline="", encoding="utf-8") as f:
                    reader = csv.DictReader(f)

                    csv_energy = 0.0
                    csv_time = 0.0

                    for row in reader:
                        energy = row.get("energy_consumed") or row.get("energy") or row.get("kwh")
                        time = row.get("duration") or row.get("time") or row.get("elapsed_time")

                        if energy is None or time is None:
                            continue

                        try:
                            csv_energy += float(energy)
                            csv_time += float(time)
                        except ValueError:
                            continue

                print(f"      Energy (kWh): {csv_energy}")
                print(f"      Duration (s): {csv_time}")

                if csv_time > 0:
                    csv_power = (csv_energy * POWER_FACTOR) / csv_time
                    print(f"      Power (W): {csv_power:.6f}")

                method_energy += csv_energy
                method_time += csv_time

            if method_time > 0:
                method_power = (method_energy * POWER_FACTOR) / method_time
            else:
                method_power = None

            print("\n    --- METHOD TOTAL ---")
            print(f"    Total Energy (kWh): {method_energy}")
            print(f"    Total Duration (s): {method_time}")
            print(f"    Method Power (W): {method_power}")

            results[dataset_name][method_name] = method_power
```

```python
df = pd.DataFrame.from_dict(results, orient="index")
df
```

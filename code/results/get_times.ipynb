```python
import pandas as pd
```

# TSC

```python
tsc_datasets = [
   'BeetleFly',
   'ACSF1',
   'CricketX',
   'CinCECGTorso',
   'LSST'
]
feat_time = pd.read_csv('../../tsc/outputs/time_multi.csv')
feat = (
    feat_time.rename(columns={'time_seconds': 'time'})
    .loc[feat_time['dataset'].isin(tsc_datasets), ['dataset', 'time']]
    .reset_index(drop=True)
)
feat
```

```python
inference_time = pd.read_csv('../../tsc/outputs/multi.csv')[['dataset', 'time']]
inference = inference_time[inference_time['dataset'].isin(tsc_datasets)].groupby('dataset', as_index=False)['time'].mean().reset_index(drop=True)
inference
```

```python
(feat.set_index('dataset')['time'] + inference.set_index('dataset')['time']).reset_index()
```
